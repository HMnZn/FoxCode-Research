# fox_ai：薄 OpenAI-compatible adapter

默认单元只演示转换，不调用付费模型。先在仓库根目录 uv sync，选择同一环境内核。真实流式实验需显式运行最后一格的 await。


```python
import os
from fox_ai.src import Model, Context, StreamOptions, UserMessage, AssistantMessage, ToolCall, ToolResultMessage, stream
from fox_ai.src.openai_provider import convert_context

context = Context([
    UserMessage("Read demo.py"),
    AssistantMessage(tool_calls=[ToolCall("read1", "Read", {"file_path": "demo.py"})]),
    ToolResultMessage("read1", "Read", "1: print(42)"),
])
convert_context(context)
```

输入通过 base_url、api_key、model 接入；没有厂商 registry。流是 TextDelta、ThinkingDelta、ToolCallDelta、Done、Error。完整工具参数只在 Done 使用。


```python
async def live():
    model = Model(os.environ["FOX_MODEL"], os.environ.get("FOX_BASE_URL", "https://api.openai.com/v1"))
    options = StreamOptions(api_key=os.environ.get("FOX_API_KEY", os.environ.get("OPENAI_API_KEY", "")))
    async for event in stream(model, Context([UserMessage("Say hello")]), options):
        print(event)

# 显式启用真实模型实验：
# await live()
```
